### Erster Test
Probiere aus, deine erste Animation laufen zu lassen. Die Animation in dem Folgendem Codeabschnitt ist eine der Standardbeispiele aus der Manim Gallerie.

```python
%%manim -qm -v WARNING SquareToCircle

class SquareToCircle(Scene):
   def construct(self):
      # Definition alles Objekte
      square = Square()
      circle = Circle()
      circle.set_fill(PINK, opacity=0.5)

      # Animationen
      self.play(Create(square))
      self.play(Transform(square, circle))
      self.wait()
```

### Manim näher kennenlernen

Wir möchten nun Manim näher kennenlernen, in dem wir gemeinsam ein Beispiel Schritt für Schritt durchgehen.

```python
%%manim -qm -v WARNING Beispiel

class Beispiel(Scene):
    def construct(self):
        # Objekte generieren
        square = Square(side_length=2, color=BLUE)
        circle = Circle(radius=1, color=GREEN)
        triangle = Triangle(color=RED)

        # Animation: Quadrat taucht auf
        self.play(Create(square))
        self.wait(0.5)

        # Objekte mit Bezug auf das Quadrat positionieren
        circle.next_to(square, RIGHT, buff=1)
        triangle.next_to(square, LEFT, buff=1)

        # Animation: Kreis und Dreieck tauchen auf
        self.play(Create(circle), Create(triangle))
        self.wait(0.5)

        # Objekte bewegen
        self.play(square.animate.rotate(PI / 4))
        self.wait(0.5)

        # Objekte gruppieren und gemeinsam bewegen
        group = VGroup(square, circle, triangle)
        self.play(group.animate.shift(UP))
        self.wait(0.5)

        # Objekte transformieren
        new_square = Square(side_length=2.83, color=YELLOW)
        new_square.move_to(square.get_center())
        self.play(Transform(circle, new_square))
        self.wait(0.5)

        # Aussehen von Objekten ändern
        self.play(triangle.animate.set_fill(RED, opacity=0.5))
        self.wait(0.5)

        # Objekte verschwinden lassen
        self.play(FadeOut(group))
        self.wait()
```

#### 1. Steigungsdreieck einer Funktion

Der folgende Code zeichnet ein Koordinatensystem und zeichnet die Funktion $f(x)=x^2$. Experiementiere mit unterschiedlichen Funktionen. Suche dir nun zwei Punkte aus, die auf der Funktion liegen. Zeichne diese Punkte ein, dann ihre Verbindung und zuletzt die Katheten des Steigunsdreiecks. Kannst du den Code so schreiben, dass man die beiden Punkte möglichst leicht ändern kann?

```python
%%manim -qm -v WARNING Steigungsdreieck

class Steigungsdreieck(Scene):
    def construct(self):
        # Zu plottende Funktion:
        def f(x):
            return x**2

        # Koordinatensystem
        axes = Axes(
            x_range=[-3, 3],
            y_range=[-0.5, 9],
            x_length=12,
            y_length=5,
        )

        # Funktionsgraph
        graph = axes.plot(f, color=BLUE)
        label = axes.get_graph_label(graph, label="x^2")

        # Animationen
        self.play(Create(axes))
        self.play(Create(graph), Write(label))
        self.wait(1)
```

#### 2. Polygon zu Kreis
Die Idee der folgenden Animation ist es, mit einem Dreieck zu starten und die Anzahl an Ecken zu erhöhen um einen Kreis anzunähern. Aktuell verwandelt der Code nur ein Dreieck in einen Kreis. Dein Ziel ist es für verschiedene n, die aktuelle Figur in ein n-Eck zu verwandeln, bevor zuletzt in einen Kreis verwandelt wird. Du kannst dazu die vorgegebene Python-Schleife nutzen.

```python
%%manim -qm -v WARNING PolygonZuKreis

class PolygonZuKreis(Scene):
    def construct(self):
        radius = 2.5

        # Dreieck
        n = 3
        polygon = RegularPolygon(n=n, radius=radius, color=BLUE)
        polygon.move_to(ORIGIN)

        self.play(Create(polygon))
        self.wait(0.5)

        # Seitenanzahl erhöhen
        for n in [4, 6, 8, 12, 24, 48]:
            # Polygon mit n Ecken erzeugen

            # Animation: Vorheriges Polygon in neues Polygon
            self.wait(0.3)

        # Kreis
        circle = Circle(radius=radius, color=GREEN)

        # Animation: Polygon zu Kreis
        self.play(Transform(polygon, circle))
        self.wait(2)
```
